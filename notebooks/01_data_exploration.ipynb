```python
import os
import pandas as pd

# Use the metabolic data folder
metabolic_folder = "../data/raw/Metabolics_Data/Metabolics_Data"

# Select the first Excel file
files = [
    f for f in os.listdir(metabolic_folder)
    if f.endswith(".xlsx")
]

print("Total Excel files:", len(files))
print("First file:", files[0])
```

```python
file_path = os.path.join(metabolic_folder, files[0])

# See what sheets are available
excel_file = pd.ExcelFile(file_path)

print("File:", files[0])
print("Sheets:")
print(excel_file.sheet_names)
```

```python
# Read the Data sheet
data_df = pd.read_excel(
    file_path,
    sheet_name="Data"
)

print("Shape:", data_df.shape)

print("\nColumn names:")
print(data_df.columns.tolist())

print("\nFirst 10 rows:")
display(data_df.head(10))
```

```python
# Inspect the Results sheet
results_df = pd.read_excel(
    file_path,
    sheet_name="Results"
)

print("Results shape:", results_df.shape)

print("\nResults columns:")
print(results_df.columns.tolist())

print("\nResults data:")
display(results_df.head(10))
```

```python
# Search the Data sheet for important metabolic variables

keywords = ["VO2", "VCO2", "METS", "Energy", "O2", "CO2"]

print("Searching the Data sheet...\n")

for keyword in keywords:
    print(f"\n--- Searching for: {keyword} ---")
    
    matches = []
    
    for row in range(data_df.shape[0]):
        for col in range(data_df.shape[1]):
            value = data_df.iloc[row, col]
            
            if pd.notna(value) and keyword.lower() in str(value).lower():
                matches.append((row, col, value))
    
    for match in matches[:20]:
        print(match)
```

```python
# Read the Data sheet WITHOUT using the first row as column names

raw_data = pd.read_excel(
    file_path,
    sheet_name="Data",
    header=None
)

print("Shape:", raw_data.shape)

# Display the first 30 rows and first 40 columns
display(raw_data.iloc[:30, :40])
```

```python
# Show every column number and the values found in the first few rows

for col in range(raw_data.shape[1]):
    values = raw_data.iloc[:4, col].tolist()
    
    print(f"Column {col}: {values}")
```

```python
# Extract the important metabolic columns

# Data starts from row 3
metabolic_measurements = raw_data.iloc[3:].copy()

# Select the important columns by their positions
metabolic_measurements = metabolic_measurements.iloc[:, [9, 14, 15, 21, 22, 23, 24]]

# Give them clean names
metabolic_measurements.columns = [
    "Time",
    "VO2",
    "VCO2",
    "VO2_Kg",
    "METS",
    "HR",
    "VO2_HR"
]

print("Shape:", metabolic_measurements.shape)

display(metabolic_measurements.head(10))
```

```python
# Convert metabolic measurement columns to numeric values

numeric_columns = [
    "VO2",
    "VCO2",
    "VO2_Kg",
    "METS",
    "HR",
    "VO2_HR"
]

for col in numeric_columns:
    metabolic_measurements[col] = pd.to_numeric(
        metabolic_measurements[col],
        errors="coerce"
    )

print(metabolic_measurements.dtypes)
```

```python
print("Number of valid measurements:")
print(metabolic_measurements[numeric_columns].count())

print("\nMissing values:")
print(metabolic_measurements[numeric_columns].isna().sum())

print("\nVO2/Kg statistics:")
print(metabolic_measurements["VO2_Kg"].describe())
```

```python
display(
    metabolic_measurements[
        ["Time", "VO2", "VCO2", "VO2_Kg", "METS", "HR"]
    ].head(20)
)
```

```python
import matplotlib.pyplot as plt

# Make sure Time is treated as numeric/time where possible
plot_df = metabolic_measurements.copy()

# Convert Time to elapsed seconds
plot_df["Time_seconds"] = pd.to_timedelta(
    plot_df["Time"].astype(str),
    errors="coerce"
).dt.total_seconds()

# Remove rows without valid values
plot_df = plot_df.dropna(
    subset=["Time_seconds", "VO2_Kg"]
)

print("Number of valid points:", len(plot_df))

# Plot VO2/Kg
plt.figure(figsize=(10, 5))

plt.plot(
    plot_df["Time_seconds"],
    plot_df["VO2_Kg"]
)

plt.xlabel("Time (seconds)")
plt.ylabel("VO2/Kg (mL/min/kg)")
plt.title("VO2/Kg During PhD13 Exopowin Trial")

plt.grid(True)
plt.show()
```

```python
print("Start time:", plot_df["Time_seconds"].min(), "seconds")
print("End time:", plot_df["Time_seconds"].max(), "seconds")

print(
    "Duration:",
    plot_df["Time_seconds"].max() - plot_df["Time_seconds"].min(),
    "seconds"
)
```

```python
end_time = plot_df["Time_seconds"].max()

final_3_min = plot_df[
    plot_df["Time_seconds"] >= end_time - 180
].copy()

print("Points in final 3 minutes:", len(final_3_min))

print("\nFinal 3-minute VO2/Kg:")
print(final_3_min["VO2_Kg"].describe())

print(
    "\nMean VO2/Kg:",
    final_3_min["VO2_Kg"].mean()
)
```

```python
# ============================================
# Calculate PhD13 standing metabolic baseline
# ============================================

# Read the standing Excel file
stand_raw = pd.read_excel(
    stand_path,
    sheet_name="Data",
    header=None
)

print("Standing raw data shape:", stand_raw.shape)


# Extract the same columns we used for the walking trial
stand_measurements = stand_raw.iloc[3:].copy()

stand_measurements = stand_measurements.iloc[
    :, [9, 14, 15, 21, 22, 23, 24]
]

stand_measurements.columns = [
    "Time",
    "VO2",
    "VCO2",
    "VO2_Kg",
    "METS",
    "HR",
    "VO2_HR"
]


# Convert measurements to numbers
numeric_columns = [
    "VO2",
    "VCO2",
    "VO2_Kg",
    "METS",
    "HR",
    "VO2_HR"
]

for col in numeric_columns:
    stand_measurements[col] = pd.to_numeric(
        stand_measurements[col],
        errors="coerce"
    )


# Convert time to seconds
stand_measurements["Time_seconds"] = pd.to_timedelta(
    stand_measurements["Time"].astype(str),
    errors="coerce"
).dt.total_seconds()


# Remove rows where time or VO2/Kg is missing
stand_plot_df = stand_measurements.dropna(
    subset=["Time_seconds", "VO2_Kg"]
).copy()


# Display information
print("Valid standing points:", len(stand_plot_df))

print(
    "Start time:",
    stand_plot_df["Time_seconds"].min(),
    "seconds"
)

print(
    "End time:",
    stand_plot_df["Time_seconds"].max(),
    "seconds"
)

print(
    "Duration:",
    stand_plot_df["Time_seconds"].max()
    - stand_plot_df["Time_seconds"].min(),
    "seconds"
)


# Calculate final 3-minute baseline
stand_end_time = stand_plot_df["Time_seconds"].max()

stand_final_3_min = stand_plot_df[
    stand_plot_df["Time_seconds"] >= stand_end_time - 180
].copy()


# Calculate baseline VO2/Kg
standing_baseline = stand_final_3_min["VO2_Kg"].mean()

print(
    "\nStanding baseline VO2/Kg:",
    standing_baseline
)

print(
    "Number of points in final 3 minutes:",
    len(stand_final_3_min)
)
```

```python
# Calculate net metabolic cost for PhD13 Exopowin

walking_cost = final_3_min["VO2_Kg"].mean()

net_metabolic_cost = walking_cost - standing_baseline

print("===================================")
print("PhD13 - Exopowin")
print("===================================")

print("Walking VO2/Kg:", walking_cost)
print("Standing baseline:", standing_baseline)
print("Net metabolic cost:", net_metabolic_cost)
```
